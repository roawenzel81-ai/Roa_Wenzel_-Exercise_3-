```python
import sys

try:
    username = str(input("Enter your name: "))
    if not username.isalpha():
        print("Name must only contain letters.")
        sys.exit(1)
    
    age = int(input("Enter your age: "))
    if age < 0:
        print("Age must be a positive integer.")
        sys.exit(1)

    with open("users.txt", "a") as file:
        file.write(f"Username: {username}, Age: {age}\n")
except ValueError:
    print("Please enter a valid age.")
    sys.exit(1)
except FileNotFoundError:
    print("The file doesn't exist. Create file first before reading.")
    sys.exit(1)
except FileExistsError:
    print("The file already exists.")
    sys.exit(1)
except SystemExit:
    print("Program exit.")
else:
    print(f"Hello, {username}! Your age is {age} years old.")
finally:
    print("System complete.")
```

```python
import sys

try:
    with open("users.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("The file doesn't exist. Create file first before reading.")
    sys.exit(1)
except FileExistsError:
    print("The file already exists.")
    sys.exit(1)
except SystemExit:
    print("Program exit.")
else:
    print("File read successfully.")
finally:
    print("System complete.")
```
